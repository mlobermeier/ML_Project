```python
import os
from pathlib import Path
import matplotlib.pyplot as plt
import matplotlib.image as mpimg

current_dir = Path().resolve()

if current_dir.name == "ML_Project":
    print(f"Current directory is already set to: {current_dir}")
else:
    for parent in current_dir.parents:
        if parent.name == "ML_Project":
            os.chdir(parent)
            print(f"Changed directory to: {os.getcwd()}")
            break
    else:
        print("ML_PROJECT directory not found in the parent directories.")
        exit(1)
```

```python


# Set folder path
figures_dir = "figures"

# List all PNG files
fig_files = [f for f in os.listdir(figures_dir) if f.endswith(".png")]

# Display all figures
for fig_file in fig_files:
    img_path = os.path.join(figures_dir, fig_file)
    img = mpimg.imread(img_path)
    plt.figure(figsize=(8, 6))
    plt.imshow(img)
    plt.axis('off')
    plt.title(fig_file)
    plt.show()
```
